```python
cars_asia = df[df.origin == "Asia"][["vehicle_weight", "model_year"]]  #q7: 1 Selecting all cars from asia and selecting columns vehicle_weight and model_year
cars_asia
```

```python
X = cars_asia.head(7).to_numpy()  # q7: 3,4 Getting 7 values and converting them to numpy arrays.
X
```

```python
X_transpose = X.T   
X_transpose
```

```python
X_transpose.shape
```

```python
XTX = np.dot(X, X_transpose)
XTX
```

```python
XTX.shape
```

```python
XTX_inverse = np.linalg.pinv(XTX)  # Invert XTX
```

```python
XTX_inverse.shape
```

```python
y = [(1100, 1300, 800, 900, 1000, 1100, 1200)]
y
```

```python
# Multiplication of the inverse of XTX with the transpose of X
X_trans_XTX_inv = np.dot(X_transpose, XTX_inverse) 
X_trans_XTX_inv.shape
#  multiply the result by y to obtain w
w = np.dot(y, X_trans_XTX_inv.T) # replaced positions to match dimensions for multiplication.
w
```

```python
sum_w = np.sum(w)    #sum of all the elements of the result
sum_w
```
